```python
import geopandas as gpd
from datetime import datetime
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from zoneinfo import ZoneInfo

from create_shadow_slide_visualization import create_shadow_slide, compute_day_positions
```

```python
BUILDINGS_PATH = "./data/buildings_campus_sp_simplified.fgb"   # <-- change this
STEP_MINUTES   = 15                              # 5 = smooth, 10 = fast preview

print("Loading buildings …")
buildings = gpd.read_file(BUILDINGS_PATH)
print(f"  {len(buildings)} buildings, CRS: {buildings.crs}\n")

start = datetime(2026, 6, 30, 0, 0, 0, tzinfo=ZoneInfo("Europe/Berlin"))
end = datetime(2026, 7, 1, 0, 0, 0, tzinfo=ZoneInfo("Europe/Berlin"))

print("Computing sun positions …")
positions = compute_day_positions(
    start, end,
    step_minutes=STEP_MINUTES,
)

create_shadow_slide(
    buildings=buildings,
    positions=positions,
    file_path="june30.gif",  # uncomment to save
    interval_ms=200,       # 200ms per frame = ~5 fps
)
```
