#### Compare: simplified height of building parts

```python
shadow_service_complete = ShadowService(None, buildings_complete_4326)
shadow_map_complete = shadow_service_complete.get_shadow_map(dt)
```

```python
shadow_map_complete.to_file("./data_shadows/complete_shadow_map_4326.fgb", driver="FlatGeobuf")
```

```python
rounded_height_service = ShadowService(None, buildings_simplified_rounded_height_4326)
shadow_map_rounded_height = rounded_height_service.get_shadow_map(dt)
```

```python
shadow_map_rounded_height.to_file("./data_shadows/rounded_height_shadow_map_4326.fgb", driver="FlatGeobuf")
```

```python
shadow_union_complete = shadow_map_complete.geometry.union_all()
shadow_union_rounded = shadow_map_rounded_height.geometry.union_all()
```

```python
total_error = shadow_union_complete.symmetric_difference(shadow_union_rounded)
print(f"Error area: {total_error.area}")
```

```python
relative_error = total_error.area / shadow_union_complete.area
print(f"Relative error: {relative_error:.2%}")
```
